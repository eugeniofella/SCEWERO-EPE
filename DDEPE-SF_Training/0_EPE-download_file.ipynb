# Step-1 Download files for precipitation

```python
url = "https://drive.google.com/file/d/1PsQQ7H9YCIIFMYfQMtC1wycsQHRflo85/view?usp=sharing"
output = "../data_epe.zip"
gdown.download(url, output, quiet=False)

with zipfile.ZipFile(output, 'r') as zip_ref:
    zip_ref.extractall("../")
```
